Задание 12

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df_clean["NA_Sales"],
    df_clean["Global_Sales"],
    alpha=0.5
)
plt.title("Связь продаж в Северной Америке и мировых продаж")
plt.xlabel("Продажи в Северной Америке (млн)")
plt.ylabel("Мировые продажи (млн)")
plt.show()
```

Одна точка означает продажи в NA и мировые продажи

---

точки вытянуты вдоль восходящей линии

---

Соответствует ли график корреляционному анализу: да, визуально сильная положительная связь
