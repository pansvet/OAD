Строим гистограмму для популярности

```python
plt.hist(df['popularity'], bins=10)
plt.xlim(0, 100)
plt.title('Популярность')
plt.xlabel('popularity')
plt.ylabel('сколько раз встречается каждое значение popularity')
plt.show()
```

На этой гистогрмме показан на оси X процент популярности, а на оси Y количество треков под каждый процент популярности. Основная часть значений находится от о до 50. Значения которые встречаются реже остальных: от 50 до 100.
